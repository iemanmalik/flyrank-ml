```python
# Week 1 Research Question

## My Lane

I selected **Lane 2: Refresh / Content Opportunity Scoring**.

This lane focuses on identifying webpages that should be reviewed first for content updates or improvements. Instead of predicting whether a page will definitely recover after being updated, the objective is to prioritize limited review resources by ranking pages according to evidence collected from search performance signals.

I chose this lane because it directly supports a real business decision: deciding which pages deserve attention first. The dataset contains many observable signals such as impressions, clicks, CTR, average position, freshness, and engagement metrics that can be combined into an evidence-based ranking.
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
## Research Question

### Decision

Which webpages should be reviewed first for content refresh?

### Unit of Analysis

One content page.

### Output

A ranked list of pages with a refresh priority score and explanation.

### Action

SEO specialists or content teams can review the highest ranked pages first to decide whether they should refresh, rewrite, expand, merge, or simply monitor the content.

### Cost of a Wrong Recommendation

False positives waste human review time on pages that do not actually need updates.

False negatives may allow important declining pages to continue losing traffic because they were not reviewed early enough.

### Why ML Helps

Many search performance signals interact with each other. Machine learning can learn patterns from multiple observable signals simultaneously and prioritize pages more effectively than simple hand-written rules while still supporting human decision making.
```
